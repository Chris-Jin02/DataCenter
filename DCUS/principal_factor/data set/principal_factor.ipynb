# Central Factor: United States data centre site selection training data set

> ** The nature of the project:** a completely new and independent data set; studies only within the United States.
> ** Scope: ** Electricity, water resources, land resources, three principal facts.
> ** Snapshot date: **2026-08-22.
> ** This version of the boundary:** Establish position tags, feature dictionaries, available data sources and research bases, without falsifying model training results.

## Core conclusions

1. Current open data can construct a training site-level feature, but none of the available data sets gives a national uniform “best location” real label.
2. Two parallel tasks were retained: a continuous recommended score of `y_score` for the regression projection (the higher, the higher the recommended); and a multi-classical projection of `y_grade ∈ {1,…,K}` (the best level 1, the worst level of K). Level `K=3/4/5` is compared as an overparameter.
3. Water, electricity, networks, land, costs and market five sub-models generate continuous sub-scoring and grade probabilities, compared with models that directly use all characteristics.
4. Each task compares multiple machine and in-depth learning models; ultimately, the optimal regression model is selected separately from the best sequence classification model, and the same algorithm is not required for both.
5. All conclusions must be cross-checked by embedded space and independent geo-test collections; rigid restraint without participation in compensatory scoring is triggered by phase-out.


# Part I: Insulated distribution of United States data centre stand-alone sites

The map is based on `im3_us_data_center_locations.gpkg` of IM3 Open Source Data Center Atlas, covering 50 states of the United States and not DC and Puerto Rico. The original data contained three parallel layers of `point`, `building`, and `campus`; there was no shared project primary key between the three layers, so that the 1,471 map records removed by `id` could not be directly interpreted as 1,471 separate parks.

## Statistical units

This figure is changed to the new ZXPROTOZX unit. The results fall into three categories:

- ** Clear Campus: ** Original `campus` Layer press `id` to reset the park.
- ** Building extrapolation site:** does not fit within the clearly defined cappas build site formed by the integration of spatial and identity information.
- **Independent point: ** Unable to be classified as a point for a clear Campus or building extrapolation site; it indicates an independent location with low confidence and does not amount to a confirmed park.

## Basis of judgement and boundary

|Priority|Basis of judgement|Default boundary|Treatment|
|---|---|---|---|
| 1 |Same `id` in the same layer|Exactly the same `id`|Keep one; currently delete 3 Campus, 2 building duplicates|
| 2 |Building centre or point landed in the campus polygon|The border is also considered internal.|Added to this explicit Campus; space contains priority over name differences|
| 3 |Recording close to Campus with identity information|Same name or the same as `ref` at a distance of 250 m; same as Oporator and 100 m|Add to Clear Campus|
| 4 |Unmatched building has a consistent identity|Same name or the same as `ref` with centre distance of 750 m; same as Oporator and 250 m|Building extrapolation sites by connecting clusters|
| 5 |Point matches the building extrapolation site|point in building; or same name/same as `ref` and 250 m; same as Operator and 100 m|Point in this extrapolation site|
| 6 |Not enough space or identity.|Beyond the above border or name, Oporator, `ref` cannot support matching|No forced integration; individual building or point remaining independent|

The approximate surface distance calculated using the WGS 84 coordinates. The use of building represents a central point rather than a complete Parcel border; buildings that cross the park boundary may be misrated. If a record meets more than one canpamus, attribution is determined “in-house priority, closer priority”.

Identity fields are first reduced to lowercase and removed spaces are accurately matched, with no vague matching of names; records from different states are not merged. Building uses a state-to-state connectivity cluster, so A and B and C enter the same extrapolation sites when B and B and C, respectively, even if the direct distance between A and C exceeds the threshold.

# Part II: Model features after alignment

In the original version, 82 electric power, 83 water resources and 106 land candidate fields were overloaded, many of which were not available nationally in a stable manner and did not enter the actual IIM3/CERF-DC site calculation. This version only retains variables that have a direct impact on the feasibility, cost or market gravity of IM3.

There are only four selection rules:

1. Must appear in the input and calculation logic of IM3 Atlas, IM3 projected dataset or CDF-DC.
2. It must be possible to collect or calculate national open spatial data, government statistics or project-specific information directly.
3. Output points, existing facility identities and future results cannot be returned to input.
4. Hard exclusions are implemented first and are not allowed to be offset by other high marks in the model.

Growth scenarios, subjective weights and aggregate size values of undefined calibres are not observationsable facts of location and therefore do not enter core characteristics.

** Maps are maintained, but the number at the state level does not enter the model. ** State, county, operator, facility name and number of existing facilities are used only for map, spatial grouping and sample inspection.

## 3.3 Strict result labels

Preserve the original state of the source and the full event time line before generating the training label. The first version of the second classification uses only the clearest two ends of the evidence, avoiding the imposition of failure on an uncertain project.

|Uniform Status|Basis of judgement|Release 1 and 2 classification processing|
|---|---|---|
| `positive_strict` |Evidence of formal authorization, authorization, commencement or operation| `y_outcome=1` |
| `negative_strict` |Dated final government veto, permit denial or planning denial, followed by no appeal| `y_outcome=0` |
| `withdrawn` |The developer withdrew the application.|separate;not automatically as location failed|
| `cancelled_other` |Cancelled for funds, clients, markets or other informal vetoes|Separate reservations|
| `delayed_or_suspended` |Postponement, suspension, litigation, morotorium impact or state of uncertainty|Retention to category II; as omission|
| `pending` |Proposed, pre-proposal or approval not yet completed|Not in Category II.|

If a project is rejected by the Planning Commission and subsequently approved on appeal, the final label of the project should be positive, provided that the early rejection event is retained. Reasons for refusal can be used as factor submodel labels or explanatory fields, not as input characteristics for predicting the same end result, otherwise the result will leak.

## 3.6 Feasibility and continuation

|Aspects|Current judgement|Basis and limits|
|---|---|---|
|Create Project Master List|High|FracTracker has a national facility schedule, longitude and extensive source links.|
|Create a strict positive sample|Medium High|Operating and Approved/Construation candidates are sufficient, but IM3 records still need independent confirmation|
|Create a strict negative sample|Medium|147 Cancelled/Suspended Candidates and Rejected/Withdrawals; final availability is dependent on re-renewal of original documents|
|Date and cause of reinstatement|Medium|DataCenterTracker and source links provide clues, but local government data are scattered|
|Acquisition of 15 decision point features|Medium to Medium|National space layer availability; projects MW, PUE, cooling and historical rates may be missing|
|First edition of Machine Learning|Medium High|Fits to table models such as Logical regression, GAM, CatBoost/ LightGBM|
|In-depth learning, first edition|Low|Strict negative samples are expected to be much less than positive samples and are not suitable for the main model|

The following are only thresholds for project planning and are not statistically effective: if **100 strict negative samples are obtained after weight, covering 20 states and at least 3 years of decision-making**, the primary surveillance experiment of national space push can be carried out; if 50-99 rigorous negative samples are matched with case-based and interpretable machine learning and conclusions are positioned as exploratory; if fewer than 50, priority is given to publishing data sets, labelling methods and case studies, without in-depth learning performance as a primary contribution. The official sample volume requirement should be calculated separately after the label distribution has been determined.
